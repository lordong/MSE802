```python
# Packages import
import numpy as np
import qiskit as qk
import qiskit_aer as qks
import qiskit.quantum_info as qkq
import qiskit.visualization as qkv
```

```python
# Task 1: Create a 2-qubit quantum circuit with a 90° phase shift and visualize it using the Q-sphere

# Crate a 2 qubit2
qc = qk.QuantumCircuit(2)

# Apply Hadamard gate to qubit 0
qc.h(0)
print(qc)

state = qkq.Statevector(qc) # State vector of 2-qubit
state.draw("qsphere")   # Draw with Q-sphere
```

```python
# Task 2: Create a circuit in qasm and import and run it in qiskit.
import qiskit.qasm2 as qkq2

# Create circuit with 2 qubits and 2 bits
print("Create circuit with 2 qubits and 2 bits.")
qc = qk.QuantumCircuit(2, 2)

qc.h(0) # Apply Hadamard gate to qubit 0
qc.cx(0, 1) # Apply cx(0, 1)
qc.measure([0, 1], [0, 1])  # Measure 2 qubits to 2 bits
print("\nThe result of circuit after apply Hadamard gate and cx(0, 1):")
print(qc)

print("\nDumps resut of circuit:")
print(qkq2.dumps(qc))

# Export to qasm file
print("\nExport to mytask.qasm file.")
qkq2.dump(qc, "mytask.qasm")

# Load qasm file
print("\nLoad mytask.qasm file and print circuit:")
circuit = qkq2.load("mytask.qasm")
print(circuit)
```

```python
# Task 3: Draw any circuit random. But don't use dump method.
import requests
import json

# Convert qasm to json code
print("Convert qasm to json code and print result:")
code = qkq2.dumps(circuit)
print(code)

# Remove qasm script: include "qelib1.inc";
print("\nRemove qasm script and print result:")
code = code[:14] + code[(14 + 22):]
print(code)

# Send code to quokka, support quokka1~5
print("\nSend code to quokka and print the random results:")
req_str_qasm = 'http://quokka5.quokkacomputing.com/qsim/qasm'
data = {
    'script': code,
    'count': 1000
}
result = requests.post(req_str_qasm, json=data)
json_obj = json.loads(result.content)
print(json_obj)
```
